## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

```python
Decision: Which pages should a content editor refresh first, out of a capacity-limited review queue?
Action: The editor updates/expands the flagged pages content (not just metadata — actual body content, given the label is about traffic decline, not just CTR).
Cost of a wrong call: Flagging a stable page wastes an editors refresh effort (Mistake A). Missing a genuinely declining page means its traffic keeps eroding, unnoticed, until it's
much harder to recover (Mistake B) — I would weight B as worse, since decline tends to compound the longer it goes unaddressed.
```
